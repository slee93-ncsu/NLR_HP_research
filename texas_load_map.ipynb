# Texas residential load breakdown

County maps and tables from the OCHRE end-use breakdown, drawn locally.
Everything runs on this machine — nothing is uploaded.

**Files needed in this folder**

- `texas_load_map.py` — the plotting code
- `tx_counties.geojson` — Texas county boundaries (254 counties, simplified)
- the breakdown CSVs written by `tnd_load_breakdown.py`

**Requires** pandas, numpy and matplotlib. `ipywidgets` is optional and only
needed for the dropdown version at the bottom.

```python
%matplotlib inline
from texas_load_map import Breakdown, GROUP_ORDER

# a glob, or a list of paths — point it at your breakdown CSVs
bd = Breakdown("county_load_breakdown*.csv")
```

## The map

`key` is one of the six groups, `"Total"`, or any single end use
(`bd.end_uses` lists them).

`metric="share"` colours by each end use's share of that county's energy.
`metric="per"` colours by kWh per modeled building, which is the one to use
when comparing magnitudes rather than composition.

```python
bd.map("Space cooling")
```

```python
bd.map("Space heating", metric="per")
```

Narrow it to one climate zone or one period.
Periods: `annual`, `summer`, `winter`, `uri`.

```python
bd.map("Water heating", zone="3A", period="winter")
```

Counties with a handful of models swing wildly — one house sets the
county value. `min_models` drops them from the colour scale.

```python
bd.map("Plug loads (MELs)", metric="per", min_models=50)
```

## Save a map for a slide

```python
bd.map("Space cooling", save="cooling_annual.png", dpi=200)
```

## Tables

```python
bd.table()                      # group shares by climate zone
```

## What is in each group

| Group | End uses |
|---|---|
| Space cooling | HVAC Cooling |
| Space heating | HVAC Heating |
| Water heating | Water Heating |
| Wet appliances | Clothes Washer, Clothes Dryer, Dishwasher |
| Plug loads (MELs) | MELs — residual plug loads ResStock does not model separately |
| Other | lighting, TV, refrigerator, freezer, cooking range, ceiling fan, pools and spas, well pump |

To regroup, edit `GROUPS` at the top of `texas_load_map.py`; anything not
named there lands in Other.

**Counties with no results**: Loving (no models sampled), Glasscock (2 models,
both vacant), Val Verde (62 occupied models, not simulated). They are drawn
grey, with Val Verde outlined in red.
